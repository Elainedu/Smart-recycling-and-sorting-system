```python
import torch
from transformers import AutoFeatureExtractor, AutoModelForImageClassification

# 加載特徵提取器和模型
feature_extractor = AutoFeatureExtractor.from_pretrained("rootstrap-org/waste-classifier")

# 手動加載模型
model = AutoModelForImageClassification.from_pretrained("rootstrap-org/waste-classifier")

# 進行預處理
from PIL import Image
import requests

# 替換為你的圖片 URL
url = "https://s.yimg.com/zp/MerchandiseImages/5BFB18B3D1-SP-13656068.jpg"
image = Image.open(requests.get(url, stream=True).raw)
inputs = feature_extractor(images=image, return_tensors="pt")

# 模型推理
outputs = model(**inputs)
logits = outputs.logits

# 獲取預測類別
predicted_class_id = logits.argmax(-1).item()
print(f"Predicted class ID: {predicted_class_id}")

# 獲取標籤映射
id2label = model.config.id2label
predicted_label = id2label[predicted_class_id]
print(f"Predicted label: {predicted_label}")
```
